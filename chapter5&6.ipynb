```python
# Step 5: 연도별 평점 변화 분석
print("\n=== Step 5: 연도별 평점 변화 ===")
year_dict = defaultdict(list)
for i in range(len(data_clean)):
    year = data_clean[i, 3]
    if year.isdigit():
        year_dict[year].append(float(data_clean[i, 2]))

year_avg = {y: np.mean(scores) for y, scores in year_dict.items()}
sorted_year = dict(sorted(year_avg.items(), key=lambda x: int(x[0])))
print("연도별 평균 평점 (일부 예시):")
for y, avg in list(sorted_year.items())[:10]:
    print(f"{y}: {avg:.1f}")

# Step 6: 결과 시각화 (선택)
print("\n=== Step 6: 시각화 ===")
years_num = np.array(list(sorted_year.keys())).astype(int)
years_avg = np.array(list(sorted_year.values()))

plt.figure(figsize=(10, 5))
plt.plot(years_num, years_avg, marker='o')
plt.title('연도별 평균 평점 변화')
plt.xlabel('연도')
plt.ylabel('평균 평점')
plt.grid(True)
plt.xticks(years_num[::5], rotation=45)  # 5년 간격
plt.tight_layout()
plt.show()
```
